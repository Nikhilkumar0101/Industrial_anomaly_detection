## Troubleshooting

- If the dataset path check fails, find the actual folder and update `--data-root`.
- If CUDA runs out of memory, retrain with `--batch-size 8`.
- If inference is slow, retrain with `--max-memory-patches 6000`.
- If imports fail after installing packages, restart the runtime and run the cells again.
- Share the exact error traceback if you need help debugging.
